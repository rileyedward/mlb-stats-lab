# Pitcher Profile

*Single-pitcher deep dive — generated by `statslab new-notebook pitcher-profile`.*

```python
# parameters
player = "Logan Gilbert"
season = None
start_date = None
end_date = None
recent_form_starts = 5
```

```python
from IPython.display import Markdown, display
from mlb_stats_lab.data import current_season, season_range, get_statcast_pitcher
from mlb_stats_lab.pitching import pitching_line, arsenal_summary, pitch_movement, location_summary, results_by_pitch, command_metrics, rolling_pitching_line
from mlb_stats_lab.pitching.splits import splits as pitching_splits
from mlb_stats_lab.viz import (pitch_movement_chart, velocity_distribution, pitch_usage_bar, pitch_usage_pie, zone_heatmap, zone_grid, splits_bar_chart, rolling_metric_line)

season = season or current_season()
if start_date is None or end_date is None:
    s, e = season_range(season)
    start_date = start_date or s
    end_date = end_date or e
display(Markdown(f'# Pitcher Profile — {player}\n*{season} season · {start_date} → {end_date}*'))
```

```python
df = get_statcast_pitcher(player, start=start_date, end=end_date)
print(f'Loaded {len(df)} pitches for {player}.')
```

## §1 Season Line + Command

```python
line = pitching_line(df)
command = command_metrics(df)
display(line.to_frame('value'))
display(command.to_frame('value'))
```

## §2 Arsenal

```python
display(arsenal_summary(df))
pitch_movement_chart(df, title=f'{player} — Arsenal Movement')
```

```python
velocity_distribution(df, kind='violin', title='Velocity Distribution')
```

```python
pitch_usage_pie(df, title='Pitch Mix')
```

## §3 Usage by Context

```python
pitch_usage_bar(df, by='vs_hand', title='Usage vs Hand')
```

```python
pitch_usage_bar(df, by='count_state', title='Usage by Count State')
```

## §4 Locations

```python
zone_heatmap(df, perspective='pitcher', metric='density', title='Pitch Locations')
```

```python
zone_grid(df, perspective='pitcher', metric='whiff_rate', title='Whiff% by Zone')
```

## §5 Results by Pitch

```python
display(results_by_pitch(df))
```

## §6 Splits

```python
for axis, title in [('vs_hand','vs Hand'),('innings','By Innings'),('times_through_order','TTO')]:
    s = pitching_splits(df, axis=axis)
    display(s)
    splits_bar_chart(s, metric='K-BB%', title=title)
```

## §7 Recent Form

```python
rolling = rolling_pitching_line(df, window=recent_form_starts, over='starts')
rolling_metric_line(rolling, metric='K%', league_average=0.225, title='Rolling K%')
```
