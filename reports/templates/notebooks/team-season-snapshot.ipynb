# Team Season Snapshot

*State of the union for a single team.*

```python
# parameters
team = "SEA"
season = None
start_date = None
end_date = None
recent_window_games = 15
```

```python
from IPython.display import Markdown, display
from mlb_stats_lab.data import current_season, season_range, get_statcast_for_team, get_standings, team_schedule, get_team_batting, get_team_pitching
from mlb_stats_lab.batting import batting_line, contact_quality, plate_discipline
from mlb_stats_lab.batting.splits import splits as batting_splits
from mlb_stats_lab.pitching import pitching_line, arsenal_summary
from mlb_stats_lab.viz import (splits_bar_chart, pitch_usage_bar, velocity_distribution, ev_la_scatter, rolling_metric_line)

season = season or current_season()
if start_date is None or end_date is None:
    s, e = season_range(season)
    start_date = start_date or s
    end_date = end_date or e
display(Markdown(f'# Team Snapshot — {team}\n*{season} season · {start_date} → {end_date}*'))
```

```python
bat_df = get_statcast_for_team(team, start=start_date, end=end_date, side='batting')
pit_df = get_statcast_for_team(team, start=start_date, end=end_date, side='pitching')
schedule = team_schedule(team, year=season)
print(f'Batting rows: {len(bat_df)}, Pitching rows: {len(pit_df)}, Schedule rows: {len(schedule)}')
```

## §1 Record

```python
display(schedule.head() if not schedule.empty else 'No schedule data.')
standings = get_standings(season)
display(standings.head(20))
```

```python
display(batting_line(bat_df).to_frame('value'))
display(contact_quality(bat_df).to_frame('value'))
display(plate_discipline(bat_df).to_frame('value'))
```

```python
splits_bar_chart(batting_splits(bat_df, axis='vs_hand'), metric='wOBA', title='Offense vs Hand')
```

```python
splits_bar_chart(batting_splits(bat_df, axis='month'), metric='wOBA', title='Offense by Month')
```

```python
ev_la_scatter(bat_df, title='Team EV / LA')
```

```python
display(pitching_line(pit_df).to_frame('value'))
display(arsenal_summary(pit_df))
```
